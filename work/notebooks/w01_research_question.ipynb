## Quick Look at the Data

I explored the starter dataset to understand whether it is suitable for the Refresh / Content Opportunity Scoring lane. The dataset contains 30,000 rows and 44 columns, providing a substantial amount of information about content performance and search metrics.

One key observation is that the correlation between search volume and impressions is only 0.001, indicating that search volume alone is not a strong indicator of how many impressions a page receives. I also observed that the median word count is very similar across different trend categories (for example, 2,909 words for declining pages and 2,847.5 words for growing pages). This suggests that content length by itself is unlikely to explain why some pages perform better than others.

These findings suggest that no single metric is enough to determine which pages should be refreshed. Because content performance depends on multiple factors, the Refresh / Content Opportunity Scoring lane is worth exploring over the next seven weeks using machine learning to combine these signals into better recommendations.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Display a few statistics from the starter dataset

# 1. Dataset size
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")

# 2. Correlation between search volume and impressions
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation (Search Volume vs Impressions): {corr:.3f}")

# 3. Median word count for declining vs growing pages
word_counts = df.groupby("trend_direction")["word_count"].median()
print("\nMedian Word Count by Trend:")
print(word_counts)
```
